## Mathematical Function


```python
x=np.array([10,20,30])
y=np.array([60,70,80])
Z= np.array([10, 20,30,np.nan, 100, np.nan]) ## normallyy  calculate korle eita nan e asbe, oijonno nan function use kora lage
add=x+y
print(add)
mul=x*y
print(mul)
sub=x-y
print(sub)
div=x/y
print(div)
print(np.reciprocal(x))
print(np.pow(x,y))
print(np.mod(x, y))

print(np.sum(x))
print(np.cumulative_sum(x))
print(np.prod(x)) ## return the product of array elements over a given axis
print(np.nanprod(Z)) ## Return the product of array elements over a given axis treating Not a Numbers (NaNs) as ones
print(np.cumulative_prod(x)) ## Return the cumulative product of elements along a given axis
print(np.cumprod(x)) ## another way to do that
print(np.cross(x, y))
print(np.diff(x)) ## describe the difference between the elements
```

## comparison operator

```python
## here  is some logical operator >, <, =>, >= , ==, != 

x=np.array([10,20,300])
y=np.array([60,70,80])
result= x<y
print(result)
print(x>y)
print(x==y)
print(np.all(x>y)) ## sobgulai howa lagbe\
print(np.any(x>y)) ## jekono ekta holey hobe
print(x!=y)
print(x>=y)

array1 = np.array([9, 12, 21])
array2 = np.array([21, 12, 9])
result=np.greater(array1, array2)
print("is greater: ", result)
result=np.less(array1,array2)
print("is less: ", result)
result=np.greater_equal(array1, array2)
print("is greater equeal: " , result)
```

## Sorting

```python
## In place Sorting
array1 = np.array([90, 12, 21])
array2 = np.array([21, 12, 9])
array3=array1.copy()
array3.sort()
print("Here is now in place sorting: ")
print(array3)
print("\n")
print("here is now copy sorting: ")
array_copy=np.sort(array1)
print(array_copy, "\n")

## 2d array sorting np.sort(a, axis= default -1 that means last axis, kind= ki dhoroner sorting, order= ki order e sort korbo, stable= ) 

## row wise horizontally
array=np.array([[100, 20,2, 35],
               [22,234,2,45]])
print("horizontally: ", np.sort(array, axis=1))
print("Vertically: ", np.sort(array, axis=0))
 ## sort the flattened array
print(np.sort(array, axis=None)     # sort the flattened array
)


print("\n another one: \n")

dtype = [('name', 'S10'), ('height', float), ('age', int)]
values = [('Arthur', 1.8, 41), ('Lancelot', 1.9, 38),
          ('Galahad', 1.7, 38)]
a = np.array(values, dtype=dtype)       # create a structured array
print(np.sort(a, order='height'))
```

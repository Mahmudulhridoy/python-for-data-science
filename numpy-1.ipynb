## Ndarray

```python
import numpy as np
# 1d Array
ara1=np.array([1,2,3,4,5]) 
print(type(ara1))
print(ara1.ndim)

#2d array
arra1=np.array([
               [1,2,3,4],
               [6,7,8,9]]
              )
print(arra1)
print(type(arra1))
print(arra1.ndim)
arra2=np.array([
    #1st floor
            [[1,2,3,4],
               [6,7,8,9]] ,
                ##2nd floor
                [[1,2,3,4],
               [6,7,8,9]],
               [[1,2,3,4],
               [6,7,8,9]]
])
print(type(arra2))
print(arra2.ndim)
```

## array Generation with random values

```python
#np.random.rand(shape)
arr=np.random.rand(2,3) # 0-->1 porjonto value dey generate kore
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#np.random.randint((start, end, (shape))
arr=np.random.randint(1,100,(2,3)) # starting point, end point, ki shape er data chacchi
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#np.random.uniform((start, end(shape))
# 
arr=np.random.uniform(50, 100, (3,4)) # 0-->1 porjonto value dey generate kore
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#np.range(start, end, step) -> similar to range function
arr = np.arange(1,10, 1)
print(arr)
print("Now Reshaping it")
matrix=arr.reshape(3,3)
print(matrix)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#np.linespace(start, end(included), koita) create values in equal distance over the end point
arr=np.linspace(1,15,10)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np. logspace(start exponent, end exponent, exponenet step)
arr=np.logspace(2,8,11) # exponent step= (start - end)/(step-1) 
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
arr=np.logspace(0, 4,2, base=2)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

# indexing and slicing 

```python
#indexing
print(ara1)
print(arra1)
print(ara1[2])
print(arra1[1][2])
arra1[1][2]=100
print(arra1[1][2])
```

```python
#slicing
#1d= arr[start: end: step]
ara1_mod=ara1[1:4]
print(ara1_mod)
```

```python
#2d slicing
# arr[row_start: row_end : step: ,column_start: column_end: step]

arra1_mod= arra1[0: 1, ]
print(arra1_mod)
arra1_mod=arra1[1: ,]
print(arra1_mod)
#gettin column
array_c=arra1[::, 0:1]

print(array_c)
array_c= arra1[: :, 1: 2]
print(array_c)


#getting a portion

portion=arra1[::, 0:2]
print(portion)
```

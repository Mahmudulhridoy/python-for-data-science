```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

df=pd.read_csv('final-employee-ds.csv')
th=np.array(df['TrainingHours'])
mean=np.mean(th)
std=np.std(th)
print(f"{"Mean:"} {std}\n {"Standard Deviation:"} {mean}")
# df.info()
```

Ans Of Q.2

```python
Department_wise_median_salary=df.groupby('Department') ['Salary'].median()
# print(Department_wise_median_salary)
df['Salary']= df['Salary'].fillna( df['Department'].map(Department_wise_median_salary))
final_df=df[(df['ExperienceYears'] >12) & (df['ProjectCount'] > 5)]
print(final_df['Salary'].sum())
# print(df.info())
# df.head()
# df
```

Ans Of Q.3

```python
filtered_df=df.loc[((df['Department']== "IT") | (df['Department']== "Finance"
 )) & ((df['Age'] >=30) & (df['Age']<=45)) & (df['PerformanceScore'] > 88)].sort_values('Salary', ascending=False)[['Name', 'Department', 'Salary', 'PerformanceScore']]
print(filtered_df)
```

Ans Of Q.4

```python
df['SalaryPerHour']=df['Salary']/(df['WorkHoursPerWeek'] * 4)
# df.head()
salary_df=df.sort_values('SalaryPerHour',ascending=False).head(5)[['Name', 'SalaryPerHour']]
print(salary_df)
```

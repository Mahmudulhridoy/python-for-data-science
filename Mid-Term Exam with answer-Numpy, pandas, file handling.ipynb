```python
import numpy as np
import pandas as pd

np.random.seed(42)

ids = np.arange(1, 11)
ages = np.random.randint(18, 60, 10)
salaries = np.random.randint(30000, 90000, 10)
departments = np.array(["HR", "IT", "Finance", "IT", "HR", "Sales", "Finance", "IT", "Sales", "HR"])

DF = pd.DataFrame({
"id": ids,
"age": ages,
"salary": salaries,
"dept": departments
})

DF.to_csv("employees.csv", index=False)
print("Sample Data Created and Saved as employees.csv")
```

Ans Of Q.1

Q1. Create NumPy arrays from the ages and salaries data generated above. Print both arrays and display their dtype, ndim, shape, and size.


```python
age=np.array(ages) ## though ages is already an array, then again create it, same for salary
salary=np.array(salaries)
print("Ages Array: ",age, "\n Dtype: ", age.dtype, " \n Shape: ", age.shape, "\n Size: ", age.size,"\n nDim: ", age.ndim )
print("Salaries Array: ", salary , "\n Dtype: ", salary.dtype, "\n Shape: ", salary.shape, "\n nDIM: ", salary.ndim, "\n size: ", salary.size)
```

Ans Of Q2

Q2. Using the salaries array, find and print the highest and lowest salary values. Also, calculate and print the average salary and age using NumPy functions.


```python
highest= np.max(salary)
lowest=np.min(salary)
print("Highest value: ", highest, "\n Lowest: ",  lowest)
average_salary=np.median(salary)
print("Average salary: ", average_salary)
average_age= np.median(age)
print("Average age: ", average_age)
```

3. From the ages array, filter and print all ages greater than 30 using a boolean condition. Then count and print how many employees are older than 30.
